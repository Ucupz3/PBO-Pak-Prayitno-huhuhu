#Praktikum 01: Mendefinisikan Kelas Abstrak Sederhana

```python
# Impor komponen yang diperlukan dari modul abc
from abc import ABC, abstractmethod

#Definisikan Kelas Abstrak
class KendaraanAbstrak(ABC):

    def __init__(self, merk):
        self.merk = merk
        print(f"Inisialisasi KendaraanAbstrak dengan merk: {self.merk}")

    def info_merk(self):
        print(f"Merk kendaraan ini adalah {self.merk}.")

    #Definisikan Metode Abstrak
    @abstractmethod
    def start_mesin(self):
        pass

    @abstractmethod
    def stop_mesin(self):
        pass

#Kdoe buat nampilin
if __name__ == "__main__":
    print("Definisi Kelas Abstrak 'KendaraanAbstrak' selesai.")

    class Mobil(KendaraanAbstrak):
        def start_mesin(self):
            print(f"Mobil {self.merk} mesin dihidupkan.")

        def stop_mesin(self):
            print(f"Mobil {self.merk} mesin dimatikan.")


    print("\nContoh definisi kelas anak 'Mobil' selesai.")

    mobil_contoh = Mobil("Toyota")
    mobil_contoh.start_mesin()
    mobil_contoh.info_merk()
    mobil_contoh.stop_mesin()
```

#Praktikum 02: Mencoba Instansiasi Kelas Abstrak

```python
from abc import ABC, abstractmethod

#Definisikan kelas abstrak
class MediaAbstrak(ABC):
    def __init__(self, judul):
        self.judul = judul
        print(f"Inisialisasi MediaAbstrak dengan judul: {self.judul}")

    @abstractmethod
    def play(self):
        """Metode abstrak untuk memulai pemutaran."""
        pass

    @abstractmethod
    def stop(self):
        """Metode abstrak untuk menghentikan pemutaran."""
        pass

# Kode nampilin eror yang di handle
if __name__ == "__main__":
    print("Mencoba membuat objek dari kelas abstrak MediaAbstrak...")

    try:
        media = MediaAbstrak("Konten Abstrak")

        print("Objek berhasil dibuat (SEHARUSNYA TIDAK TERJADI).")
        media.play()

    except TypeError as e:
        # Menangkap dan menampilkan error TypeError
        print(f"\nGAGAL membuat objek!")
        print(f"Error yang muncul (sesuai harapan): {e}")
        print("\nIni membuktikan bahwa kelas abstrak tidak bisa diinstansiasi")
        print("jika masih memiliki metode abstrak yang belum diimplementasikan.")
```

#Praktikum 03: Membuat Subclass Konkret - Alat Pembayaran

```python
from abc import ABC, abstractmethod
import random

# formar ruypiah tanpa local sistem
def format_rupiah(angka):
    return "Rp {:,.0f}".format(angka).replace(',', '.')

#Definisikan Kelas Abstrak
class AlatPembayaranAbstrak(ABC):
    def __init__(self, nama_metode):
        self.nama_metode = nama_metode
        print(f"Inisialisasi alat pembayaran: {self.nama_metode}")

    def info(self):
        print(f"Metode Pembayaran: {self.nama_metode}")

    @abstractmethod
    def proses_pembayaran(self, jumlah):
        pass

#Implementasi Subclass Konkret

class KartuKredit(AlatPembayaranAbstrak):
    def __init__(self, nomor_kartu, nama_pemilik):
        super().__init__("Kartu Kredit")
        self.nomor_kartu = nomor_kartu[-4:]
        self.nama_pemilik = nama_pemilik
        print(f" -> Kartu Kredit ************{self.nomor_kartu} ({self.nama_pemilik}) siap.")

    def proses_pembayaran(self, jumlah):
        print(f" Memproses pembayaran {format_rupiah(jumlah)} via Kartu Kredit ************{self.nomor_kartu}...")
        berhasil = random.choice([True, False])
        if berhasil:
            print(" Pembayaran Kartu Kredit Berhasil.")
            return True
        else:
            print(" Pembayaran Kartu Kredit Gagal.")
            return False

class DompetDigital(AlatPembayaranAbstrak):
    def __init__(self, nomor_telepon, nama_provider):
        super().__init__(f"Dompet Digital ({nama_provider})")
        self.nomor_telepon = nomor_telepon
        self._saldo = random.randint(50000, 500000)
        print(f" -> Dompet Digital {self.nomor_telepon} siap (Saldo: {format_rupiah(self._saldo)}).")

    def proses_pembayaran(self, jumlah):
        print(f" Memproses pembayaran {format_rupiah(jumlah)} via Dompet Digital {self.nomor_telepon}...")
        if jumlah <= self._saldo:
            self._saldo -= jumlah
            print(" Pembayaran Dompet Digital Berhasil.")
            print(f" Saldo tersisa: {format_rupiah(self._saldo)}")
            return True
        else:
            print(" Pembayaran Dompet Digital Gagal (Saldo tidak mencukupi).")
            print(f" Saldo saat ini: {format_rupiah(self._saldo)}")
            return False

# Kode nampilin
if __name__ == "__main__":
    print("\n--- Simulasi Transaksi ---")

    kartu_bca = KartuKredit("1234-5678-9012-3456", "Budi Cahyono")
    gopay = DompetDigital("08123456789", "GoPay")

    print("\nMelakukan Pembayaran:")

    kartu_bca.info()
    kartu_bca.proses_pembayaran(150000)

    print("-" * 30)

    gopay.info()
    gopay.proses_pembayaran(75000)
```

#Praktikum 04: Menggunakan Kelas Abstrak untuk Polimorfisme - Dokumen


```python
from abc import ABC, abstractmethod
import time

#Definisi Kelas Abstrak
class DokumenAbstrak(ABC):
    def __init__(self, nama_file):
        self.nama_file = nama_file
        print(f"Inisialisasi Dokumen: {self.nama_file}")

    def info_file(self):
        """Metode konkret untuk menampilkan informasi file."""
        print(f"Nama File: {self.nama_file}")

    @abstractmethod
    def cetak(self):
        """Metode abstrak untuk mencetak dokumen."""
        pass

    @abstractmethod
    def simpan(self):
        """Metode abstrak untuk menyimpan dokumen."""
        pass

#Subclass Konkret 1: DokumenTeks
class DokumenTeks(DokumenAbstrak):
    def __init__(self, nama_file, isi_teks):
        super().__init__(nama_file)
        self.isi_teks = isi_teks
        print(" -> Tipe: Dokumen Teks")

    def cetak(self):
        print(f"Mencetak Dokumen Teks '{self.nama_file}':")
        print("-" * 20)
        print(self.isi_teks)
        print("-" * 20)

    def simpan(self):
        print(f"Menyimpan Dokumen Teks '{self.nama_file}' ke penyimpanan...")
        time.sleep(0.3)
        print(" -> Simpan Berhasil.")

#Subclass Konkret 2: Spreadsheet
class Spreadsheet(DokumenAbstrak):
    def __init__(self, nama_file, baris, kolom):
        super().__init__(nama_file)
        self.baris = baris
        self.kolom = kolom
        print(" -> Tipe: Spreadsheet")

    def cetak(self):
        print(f"Mencetak Spreadsheet '{self.nama_file}' dengan ukuran {self.baris}x{self.kolom}.")

    def simpan(self):
        print(f"Menyimpan Spreadsheet '{self.nama_file}' ke format .xlsx...")
        time.sleep(0.3)
        print(" -> Simpan Berhasil.")

#Fungsi Polimorfik
def proses_semua_dokumen(daftar_dokumen):
    """
    Fungsi ini mendemonstrasikan polimorfisme.
    Ia memanggil metode yang sama pada objek yang berbeda.
    """
    print("\n--- MEMULAI PROSES BATCH DOKUMEN ---")
    for doc in daftar_dokumen:
        print(f"\n[Sistem] Memproses: {doc.nama_file}")
        doc.cetak()
        doc.simpan()
    print("\n--- SEMUA PROSES SELESAI ---")

# Kode buat nampilin
if __name__ == "__main__":
    # contoh beberpa objek
    dok1 = DokumenTeks("laporan_akhir.txt", "Isi: Semua tugas telah diselesaikan.")
    dok2 = Spreadsheet("rekap_nilai.xlsx", 50, 10)
    dok3 = DokumenTeks("memo.txt", "Jangan lupa rapat jam 10 pagi.")

    # Memasukkan objek ke dalam list (koleksi polimorfik)
    antrean_dokumen = [dok1, dok2, dok3]

    # Menjalankan fungsi polimorfik
    proses_semua_dokumen(antrean_dokumen)
```

#Praktikum 05: Kelas Abstrak dengan Metode Konkret dan Abstrak

```python
from abc import ABC, abstractmethod
import math

#Definisikan Kelas Abstrak
class BangunDatarAbstrak(ABC):
    def __init__(self, nama):
        self._nama = nama
        print(f"Inisialisasi BangunDatarAbstrak: {self._nama}")

    #Metode Konkret

    @property
    def nama(self):
        """Getter untuk nama bangun datar."""
        return self._nama

    def info_lengkap(self):
        """Metode konkret yang memanggil metode abstrak di dalamnya."""
        print(f"\n--- Info Lengkap: {self.nama} ---")
        try:
            luas = self.hitung_luas()
            print(f" Luas      : {luas:.2f}")

            keliling = self.hitung_keliling()
            print(f" Keliling  : {keliling:.2f}")
        except Exception as e:
            print(f" Terjadi kesalahan saat menghitung: {e}")
        print("-" * (len(self.nama) + 18))

    #Metode Abstrak

    @abstractmethod
    def hitung_luas(self):
        """Metode abstrak untuk menghitung luas."""
        pass

    @abstractmethod
    def hitung_keliling(self):
        """Metode abstrak untuk menghitung keliling."""
        pass

#Implementasi Subclass Konkret

class Lingkaran(BangunDatarAbstrak):
    def __init__(self, radius):
        super().__init__("Lingkaran")
        if radius < 0:
            raise ValueError("Radius tidak boleh negatif")
        self.radius = radius
        print(f" -> Lingkaran dibuat (Radius: {self.radius}).")

    # metode abstrak hitung_luas
    def hitung_luas(self):
        return math.pi * (self.radius ** 2)

    # metode abstrak hitung_keliling
    def hitung_keliling(self):
        return 2 * math.pi * self.radius

# Kode buat nampilin (Main Program)
if __name__ == "__main__":
    print("Membuat objek Lingkaran...")
    lingkaran_A = Lingkaran(10)

    print(f"\nNama bangun (via property): {lingkaran_A.nama}")

    lingkaran_A.info_lengkap()
```

#Praktikum 06: Properti Abstrak - Elemen Grafis


```python
from abc import ABC, abstractmethod

#Definisikan Kelas Abstrak
class ElemenGrafisAbstrak(ABC):
    def __init__(self, nama):
        self._nama = nama
        print(f"Inisialisasi Elemen Grafis: {self._nama}")

    #Properti Abstrak
    @property
    @abstractmethod
    def posisi(self):
        """Properti abstrak untuk mendapatkan koordinat (x, y)."""
        pass

    @abstractmethod
    def gambar(self):
        """Metode abstrak untuk menggambar elemen."""
        pass

    def tampilkan_status(self):
        """Metode konkret untuk menampilkan informasi elemen."""
        print(f"Elemen: {self._nama} | Posisi saat ini: {self.posisi}")

# Implementasi Subclass Konkret

class Titik(ElemenGrafisAbstrak):
    def __init__(self, x, y):
        super().__init__("Titik")
        self._x = x
        self._y = y


    @property
    def posisi(self):

        return (self._x, self._y)

    def gambar(self):
        print(f"Menggambar titik tunggal di koordinat {self.posisi}")

class Garis(ElemenGrafisAbstrak):
    def __init__(self, x1, y1, x2, y2):
        super().__init__("Garis")
        self._start = (x1, y1)
        self._end = (x2, y2)


    @property
    def posisi(self):

        mid_x = (self._start[0] + self._end[0]) / 2
        mid_y = (self._start[1] + self._end[1]) / 2
        return (mid_x, mid_y)

    def gambar(self):
        print(f"Menggambar garis dari {self._start} ke {self._end}")

# kdoe buat nampilin
if __name__ == "__main__":
    print("--- Membuat Objek Grafis ---")

    obj_titik = Titik(10, 20)
    obj_garis = Garis(0, 0, 100, 100)

    print("\n--- Output Program ---")

    # Memanggil method pakai abstrajk
    obj_titik.tampilkan_status()
    obj_titik.gambar()

    print("-" * 30)

    obj_garis.tampilkan_status()
    obj_garis.gambar()
```

#Penugasan

```python
from abc import ABC, abstractmethod

#Buat Kelas Abstrak (Senjata)
class Senjata(ABC):
    def __init__(self, nama: str):
        self.nama = nama

    @abstractmethod
    def serang(self):
        """Metode abstrak untuk menyerang, harus diimplementasikan oleh subclass."""
        pass

    @property
    @abstractmethod
    def sisa_penggunaan(self):
        """Properti abstrak untuk amunisi atau daya tahan."""
        pass

    @sisa_penggunaan.setter
    @abstractmethod
    def sisa_penggunaan(self, nilai):
        """Setter abstrak untuk mengubah sisa amunisi/daya tahan."""
        pass

    def info_nama(self):
        """Metode konkret (opsional) untuk perilaku umum."""
        return f"--- Senjata: {self.nama} ---"


#Buat Kelas Anak Konkret 1 (Pedang)
class Pedang(Senjata):
    def __init__(self, nama: str, panjang_bilah: float):
        super().__init__(nama)
        self.panjang_bilah = float(panjang_bilah)
        self._daya_tahan = 100

    def serang(self):
        print(f"Pedang '{self.nama}' menebas dengan bilah sepanjang {self.panjang_bilah} cm!")
        self._daya_tahan -= 2

    @property
    def sisa_penggunaan(self):
        return self._daya_tahan

    @sisa_penggunaan.setter
    def sisa_penggunaan(self, nilai):
        self._daya_tahan = nilai


#Buat Kelas Anak Konkret 2 (Panah)
class Panah(Senjata):
    def __init__(self, nama: str, jumlah_anak_panah: int):
        super().__init__(nama)
        self.jumlah_anak_panah = jumlah_anak_panah

    def serang(self):
        if self.jumlah_anak_panah > 0:
            self.jumlah_anak_panah -= 1
            print(f"Panah '{self.nama}' melesat! Sisa anak panah: {self.jumlah_anak_panah}.")
        else:
            print(f"Amunisi Panah '{self.nama}' habis!")

    @property
    def sisa_penggunaan(self):
        return self.jumlah_anak_panah

    @sisa_penggunaan.setter
    def sisa_penggunaan(self, nilai):
        self.jumlah_anak_panah = nilai


#Buat kdoe nampilin
if __name__ == "__main__":
    print("=== PENGUJIAN KELAS SENJATA ===")

    #Buat objek dari kelas Pedang dan Panah
    pedang_excalibur = Pedang("Excalibur", 120.5)
    panah_elven = Panah("Elven Bow", 2)

    #Panggil metode serang() beberapa kali pada objek Panah
    print(panah_elven.info_nama())
    panah_elven.serang() # Sisa: 1
    panah_elven.serang() # Sisa: 0
    panah_elven.serang() # Habis

    #Panggil metode serang() pada objek Pedang
    print(pedang_excalibur.info_nama())
    pedang_excalibur.serang()
    pedang_excalibur.serang()

    #Akses properti abstrak (getter) dan manfaatkan setter
    print("\n--- Status Properti ---")
    print(f"Daya tahan {pedang_excalibur.nama} saat ini: {pedang_excalibur.sisa_penggunaan}%")
    print(f"Amunisi {panah_elven.nama} saat ini: {panah_elven.sisa_penggunaan} buah")

    print("\n--- Mengisi Ulang Amunisi Panah ---")
    panah_elven.sisa_penggunaan = 5 # Memanggil setter
    print(f"Amunisi {panah_elven.nama} setelah diisi ulang: {panah_elven.sisa_penggunaan} buah")

    #Coba buat objek dari kelas abstrak (akan menghasilkan TypeError)
    print("\n--- Uji Instansiasi Kelas Abstrak ---")
    try:
        senjata_misterius = Senjata("Senjata Random")
    except TypeError as e:
        print(f"[Error Berhasil Ditangkap]: Tidak dapat membuat instansiasi dari kelas abstrak Senjata.\nDetail: {e}")
```
