#Praktikum 01 Membuat dan Menggunakan Fungsi

```python
# Bagian 1: Fungsi Built-in
print("Coontoh fungsi built-in")

# Fungsi len()
kata = "Pemrograman"
panjang_kata = len(kata)
print(f"Panjang kata '{kata}' adalah: {panjang_kata}")

# Fungsi max()
angka = [10, 5, 30, 40, 25]
nilai_max = max(angka)
print(f"Nilai maksimum dalam daftar {angka} adalah: {nilai_max}")

# Bagian 2: Fungsi User-Defined
# Fungsi dengan satu parameter
def cetak_kuadrat(angka): #

    print(f"Kuadrat dari {angka} adalah: {angka ** 2}")

# Fungsi dengan beberapa parameter
def hitung_luas_persegi_panjang(panjang, lebar):
    # Menghitung luas persegi panjang
    return panjang * lebar

# Fungsi dengan beberapa tipe parameter
def info_mahasiswa(nama, umur, ipk):
    # Mencetak informasi mahasiswa
    print(f"Nama: {nama}, Umur: {umur}, IPK: {ipk}")

# Fungsi tanpa return value
def sapa_pengguna(nama):
    # Mencetak sapaan tanpa mengembalikan nilai
    print(f"Halo, {nama}! Selamat datang di dunia Python.")

# Fungsi dengan return value
def hitung_keliling_persegi(sisi):
    # Mengembalikan keliling persegi
    return 4 * sisi

# Bagian 3: Pemanggilan Fungsi
cetak_kuadrat(5)

luas = hitung_luas_persegi_panjang(10, 5)
print(f"Luas persegi panjang: {luas}")

info_mahasiswa("Bubud", 22, 3.8)

sapa_pengguna("Agung")

keliling = hitung_keliling_persegi(5)
print(f"Keliling persegi dengan sisi 5 adalah: {keliling}")
```

#Praktikum 3: Membuat Kelas Sederhana

```python
# --- Bagian 1: Kelas Buku ---
# Kelas Buku untuk merepresentasikan buku di perpustakaan
class Buku:
    def __init__(self, judul, pengarang, tahun_terbit):
        self.judul = judul
        self.pengarang = pengarang
        self.tahun_terbit = tahun_terbit
        self.status = "Tersedia"  # Status defaultnya Tersedia

    def tampilkan_info(self):
        print(f"Judul: {self.judul}")
        print(f"Pengarang: {self.pengarang}")
        print(f"Tahun Terbit: {self.tahun_terbit}")
        print(f"Status: {self.status}")

    def pinjam(self):
        if self.status == "Tersedia":
            self.status = "Dipinjam"
            print(f"Buku '{self.judul}' telah dipinjam.")
        else:
            print(f"Buku '{self.judul}' sedang dipinjam.")

    def kembalikan(self):
        if self.status == "Dipinjam":
            self.status = "Tersedia"
            print(f"Buku '{self.judul}' telah dikembalikan.")
        else:
            print(f"Buku '{self.judul}' tidak sedang dipinjam.")

# --- Bagian 2: Kelas Mahasiswa ---
class Mahasiswa:
    # Konstruktor (__init__) untuk menginisialisasi atribut objek
    def __init__(self, nama, nim, umur):
        self.nama = nama
        self.nim = nim
        self.umur = umur

    # Cara menampilkan informasi mahasiswa
    def tampilkan_info(self):
        print(f"Nama: {self.nama}")
        print(f"NIM: {self.nim}")
        print(f"Umur: {self.umur} tahun")

    # Cara merubah umur mahasiswa
    def ubah_umur(self, umur_baru):
        self.umur = umur_baru

# --- Bagian 3: Membuat Objek dan Menjalankan ---
# Membuat objek dari kelas Buku
buku1 = Buku("Pemrograman Python 1", "Carl Joseph", 2018)
buku2 = Buku("Data Science untuk Pemula", "Kenny Smith", 2020)

# Menggunakan metode objek Buku
buku1.tampilkan_info()
buku2.pinjam()
buku2.kembalikan()

print("\n" + "="*30 + "\n")

# Membuat objek (instance) dari kelas Mahasiswa
mhs1 = Mahasiswa("Andi", "12345", 20)
mhs2 = Mahasiswa("Budi", "67890", 22)

# Menggunakan metode dari kelas Mahasiswa
mhs1.tampilkan_info()
mhs1.ubah_umur(21)

print("\nSetelah mengubah umur mahasiswa 1:")
mhs1.tampilkan_info()
```

#Praktikum 4: Membuat dan Menggunakan Method dalam Kelas

```python
# 1. Definisi Kelas Mobil
class Mobil:
    def __init__(self, merk, warna, tahun, harga):
        # Atribut objek
        self.merk = merk
        self.warna = warna
        self.tahun = tahun
        self.harga = harga

    # Metode tanpa return value (hanya mencetak)
    def tampilkan_info(self):
        print(f"Mobil {self.merk} berwarna {self.warna}, tahun {self.tahun}, harga: Rp {self.harga}")

    # Metode dengan satu parameter untuk menghitung diskon
    def diskon(self, persen_diskon):
        diskon_harga = self.harga * (persen_diskon / 100)
        harga_setelah_diskon = self.harga - diskon_harga
        print(f"Harga setelah diskon {persen_diskon}%: Rp {harga_setelah_diskon}")

    # Metode dengan return value untuk menghitung usia mobil
    def hitung_usia(self, tahun_sekarang):
        usia = tahun_sekarang - self.tahun
        return usia

    # Metode dengan beberapa parameter
    def perbarui_harga(self, harga_baru, tahun_baru):
        self.harga = harga_baru
        self.tahun = tahun_baru
        print(f"Harga dan tahun mobil {self.merk} diperbarui menjadi Rp {self.harga} dan tahun {self.tahun}")

# 2. Implementasi Objek
# Objek mobil
mobil1 = Mobil("Porshce", "Biru", 2015, 300000000)
mobil2 = Mobil("Ferari", "Merah", 2018, 250000000)

# Info
mobil1.tampilkan_info()
mobil2.tampilkan_info()

# Menggunakan metode diskon
mobil1.diskon(10)
mobil2.diskon(15)

# Menggunakan metode dengan return value
usia_mobil1 = mobil1.hitung_usia(2025)
usia_mobil2 = mobil2.hitung_usia(2025)
print(f"Usia mobil1 pada tahun 2025: {usia_mobil1} tahun")
print(f"Usia mobil2 pada tahun 2025: {usia_mobil2} tahun")

# Menggunakan metode dengan beberapa parameter
mobil1.perbarui_harga(280000000, 2022)
```

#Penugasan membuat Penjualan HP

```python
# File utama: main_penjualan.py
import toko_hp_mas_amba

def main():
    print("=== SISTEM PENJUALAN HP ===")

    # 1. KOnsep kelas
    hp1 = toko_hp_mas_amba.Handphone("Samsung", "Galaxy S45", 15000000)
    hp2 = toko_hp_mas_amba.Handphone("iPhone", "18 Pro", 20000000)

    # Menampilkan barang
    print(f"1. {hp1.tampilkan_detail()}")
    print(f"2. {hp2.tampilkan_detail()}")
    print("-" * 30)

    # 2. Input manual untuk simulasi pembelinan
    pilihan = input("Pilih HP yang ingin dibeli (1/2): ")
    jumlah = int(input("Masukkan jumlah pembelian: "))

    if pilihan == "1":
        item = hp1
    elif pilihan == "2":
        item = hp2
    else:
        print("Pilihan tidak valid.")
        return

    # 3. Menggunakan Modularisasi Fungsi
    total = toko_hp_mas_amba.hitung_total_bayar(item.harga, jumlah)

    print(f"\nRingkasan Pembelian:")
    print(f"Produk      : {item.merk} {item.model}")
    print(f"Jumlah      : {jumlah} unit")
    print(f"Total Bayar : Rp {total:,.0f}")
    print("===========================")

if __name__ == "__main__":
    main()
```
